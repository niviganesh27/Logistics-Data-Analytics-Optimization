# Week 3 – Advanced Data Analysis and Visualization

This notebook explores shipment performance, delivery time, route-related variables,
and operational factors using descriptive statistics, correlation analysis, and visualizations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../data/processed/cleaned_logistics_data.csv")
df.head()
```

## 1. Dataset Overview

```python
print("Shape:", df.shape)
print("\nData types:")
print(df.dtypes)

print("\nDescriptive statistics:")
display(df.describe())

print("\nCategory frequencies:")
for col in ["Traffic_Level", "Weather", "Vehicle_Type", "Delivery_Priority"]:
    print(f"\n{col}")
    print(df[col].value_counts())
```

## 2. Correlation Analysis

```python
numeric_cols = [
    "Delivery_Time_Hours", "Distance_km", "Package_Weight_kg",
    "Stops_Count", "Warehouse_Delay_Min", "Driver_Experience_Years"
]

corr = df[numeric_cols].corr()
display(corr)
```

## 3. Delivery Time Distribution

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Delivery_Time_Hours"], bins=25)
plt.title("Distribution of Delivery Time")
plt.xlabel("Delivery Time (Hours)")
plt.ylabel("Number of Shipments")
plt.tight_layout()
plt.savefig("../visualizations/delivery_time_distribution.png", dpi=150)
plt.show()
```

## 4. Distance vs Delivery Time

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    df["Distance_km"],
    df["Delivery_Time_Hours"],
    s=10,
    alpha=0.35
)
plt.title("Distance vs Delivery Time")
plt.xlabel("Distance (km)")
plt.ylabel("Delivery Time (Hours)")
plt.tight_layout()
plt.savefig("../visualizations/distance_vs_delivery_time.png", dpi=150)
plt.show()
```

## 5. Vehicle Type Comparison

```python
avg_vehicle = (
    df.groupby("Vehicle_Type")["Delivery_Time_Hours"]
    .mean()
    .sort_values()
)

plt.figure(figsize=(8, 5))
plt.bar(avg_vehicle.index, avg_vehicle.values)
plt.title("Average Delivery Time by Vehicle Type")
plt.xlabel("Vehicle Type")
plt.ylabel("Average Delivery Time (Hours)")
plt.tight_layout()
plt.savefig("../visualizations/vehicle_type_delivery_time.png", dpi=150)
plt.show()

display(avg_vehicle.to_frame("Average_Delivery_Time_Hours"))
```

## 6. Traffic Level Analysis

```python
order = ["Low", "Medium", "High"]
avg_traffic = (
    df.groupby("Traffic_Level")["Delivery_Time_Hours"]
    .mean()
    .reindex(order)
)

plt.figure(figsize=(8, 5))
plt.bar(avg_traffic.index, avg_traffic.values)
plt.title("Average Delivery Time by Traffic Level")
plt.xlabel("Traffic Level")
plt.ylabel("Average Delivery Time (Hours)")
plt.tight_layout()
plt.savefig("../visualizations/traffic_level_delivery_time.png", dpi=150)
plt.show()
```

## 7. Correlation Heatmap

```python
corr = df[numeric_cols].corr()

plt.figure(figsize=(8, 6))
plt.imshow(corr, aspect="auto")
plt.colorbar(label="Correlation")
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45, ha="right")
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Logistics Variable Correlation")
plt.tight_layout()
plt.savefig("../visualizations/correlation_heatmap.png", dpi=150)
plt.show()
```
